```python
df["family"]= df["sibsp"]+ df["parch"]
df.head()
```

```python
def family_type(number):
    if number==0:
        return "Alone"
    elif number>0 and number<=4:
        return "Medium"
    else:
        return "Large"
df["family_type"]= df["family"].apply(family_type)
```

```python
df.tail()
```

```python
df.drop(columns=["sibsp","parch","family","alive","alone","class","embark_town"], inplace=True)
```

```python
df.tail()
```

```python
pd.crosstab(df["family_type"],df["survived"]).apply(lambda r : round((r/r.sum())*100,1), axis=1)
```

```python
sns.countplot(x="survived", hue="family_type", data=df, palette="magma")
```

```python
Q1=np.percentile(df["fare"],25)
Q3=np.percentile(df["fare"],75)
IQR=Q3-Q1
Q_min= Q1- 1.5*IQR
Q_max=Q3 + 1.5*IQR

df=df[(df["fare"]>Q_min) & (df["fare"]<Q_max)]
```

```python
df=pd.get_dummies(data=df, columns=["sex","pclass","family_type","embarked"],drop_first=True)
```

```python
df[df.select_dtypes(bool).columns] = df.select_dtypes(bool).astype(int)
```
